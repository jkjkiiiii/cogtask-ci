# Quickstart: nested task-budget evaluation

CogTask-CI uses entirely synthetic labels and text here. These results are not clinical evidence. Install the package and notebook extra before running. No real participant data is downloaded.

```python
from pathlib import Path
import tempfile
import csv
from IPython.display import Image, display
from cogtask_ci import Config, evaluate, validate_dataset
from cogtask_ci.demo import generate_demo
# A fresh directory makes every execution safe to repeat.
work = Path(tempfile.mkdtemp(prefix="cogtask_ci_notebook_"))
print("Local synthetic outputs:", work)
```

```python
source = generate_demo(work / "synthetic.jsonl", subjects=36)
validate_dataset(source)
```

```python
metrics = evaluate(source, work / "run", Config(outer_folds=3, inner_folds=2, bootstrap=100))
metrics
```

```python
display(Image(filename=str(work / "run" / "task_budget.png")))
print("HTML report:", work / "run" / "report.html")
```

## Interpretation

Read differences and paired intervals, not only the largest point estimate. The redundant task shares its artificial signal with task A; task B carries a separately sampled signal. Finite samples need not produce monotone curves. Quantity is constant by design. Fixed-token prefix results depend on the intentionally chosen token order. Bootstrap intervals condition on realized fits and folds; they do not include development uncertainty. All task budgets within an arm share participants. Matched arms are compared on the same eligible population.
